# **Задание №6. Разработка UML-диаграмм для проектирования информационной системы**

## **Введение**

UML-моделирование является одним из основных инструментов в разработке информационных систем, обеспечивая визуальное представление архитектуры, поведения и структуры системы. В рамках управления проектами создания информационных систем грамотное проектирование с использованием UML-диаграмм позволяет существенно снизить риски и обеспечить четкое понимание разрабатываемой системы всеми участниками проекта.

## **Цель задания**

На основе ранее разработанной концептуальной основы дипломного проекта (Задание №1), анализа существующих решений (Задание №2), иерархической структуры работ (Задание №3) и спецификации требований к системе (Задание №4), спроектировать основные UML-диаграммы, необходимые для моделирования вашей информационной системы.

## **Краткая теория**

### **1. Диаграмма классов**

**Диаграмма классов** представляет статическую структуру системы, отображая классы, их атрибуты, методы и взаимосвязи между ними. Это базовая диаграмма объектно-ориентированного моделирования, которая описывает типы объектов в системе и различные виды статических отношений между ними. Применяется на этапе проектирования для определения архитектуры программного обеспечения.

image.png

### **2. Диаграмма вариантов использования**

**Диаграмма вариантов использования** визуализирует функциональные требования к системе с точки зрения внешних пользователей (акторов). Она определяет, какие функции система должна выполнять для каждой категории пользователей, не вдаваясь в детали реализации. Эта диаграмма служит инструментом коммуникации между разработчиками и заказчиками на ранних этапах проектирования.


image.png

### **3. Диаграмма последовательности**

**Диаграмма последовательности** иллюстрирует временную последовательность взаимодействия объектов в рамках конкретного сценария работы системы. Она показывает, какие сообщениями обмениваются объекты и в каком порядке это происходит. Используется для моделирования логики алгоритмов и детализации вариантов использования.


image.png

### **4. Диаграмма состояний**

**Диаграмма состояний** моделирует возможные состояния объекта определенного класса и переходы между этими состояниями в ответ на события. Отображает жизненный цикл объекта, показывая, какие события вызывают изменение состояния и какие действия выполняются при переходе. Применяется для описания поведения объектов, имеющих различные состояния и сложную логику переходов.


image.png

### **5. Диаграмма деятельности**

**Диаграмма деятельности** представляет последовательность действий, решений и параллельных процессов, образующих определенный рабочий процесс. Позволяет моделировать как вычислительные, так и организационные процессы (бизнес-процессы). Эффективна для спецификации, конструирования и документирования динамического поведения системы.

image.png

### **6. Диаграмма компонентов**

**Диаграмма компонентов** показывает организацию и зависимости между компонентами программного обеспечения, включая исходный код, бинарные файлы и исполняемые модули. Отображает структурные отношения между элементами системы, их интерфейсы и зависимости. Используется при проектировании модульной архитектуры системы.

image.png

### **7. Диаграмма развертывания**

**Диаграмма развертывания** описывает физическую инфраструктуру, на которой выполняется система — аппаратные средства, узлы выполнения и их взаимосвязи. Показывает, как программные компоненты распределяются по физическим вычислительным ресурсам. Применяется при планировании развертывания системы в целевой среде.


image.png

### **8. Диаграмма коммуникации**












**Диаграмма коммуникации** (ранее называлась диаграммой сотрудничества) фокусируется на структурных аспектах взаимодействия объектов. В отличие от диаграммы последовательности, она акцентирует внимание не на временной последовательности, а на организации объектов и связях между ними. Используется, когда важно понять структуру взаимодействующих объектов.

   - Диаграмма деятельности (Activity Diagram) для одного из бизнес-процессов

4. **Выявление отношений**:

4. **Разделение ответственности**:
   - Какие классы отвечают за какую функциональность?
   - Правильно ли распределены обязанности между участниками?

5. **Детализация сообщений**:
   - Какие параметры передаются в сообщениях?
   - Какие возвращаемые значения ожидаются?
   - Какие альтернативные пути выполнения возможны?

### **4. Диаграмма состояний**

1. **Выбор объекта для моделирования**:

   - Какой объект проходит через различные состояния?

2. **Определение состояний**:
   - В каких различных состояниях может находиться объект?
   - Какие состояния являются взаимоисключающими?

3. **Идентификация событий и триггеров**:
   - Какие события вызывают переход из одного состояния в другое?
   - Какие условия должны выполняться для перехода?

4. **Определение действий**:
   - Какие действия выполняются при входе в состояние?
   - Какие действия выполняются при выходе из состояния?
   - Какие действия выполняются во время перехода?

5. **Выявление начального и конечного состояний**:
   - С какого состояния начинается жизненный цикл объекта?
   - Какие состояния являются терминальными?

### **5. Диаграмма деятельности**


1. **Определение процесса**:

   - Какие действия выполняются на каждом шаге?

3. **Определение порядка действий**:
   - Какова последовательность выполнения действий?

4. **Идентификация точек принятия решений**:
   - Где процесс может пойти по разным путям?
   - Какие условия определяют выбор пути?

5. **Распределение ответственности**:
   - Кто отвечает за выполнение каждого действия?
   - Как организовать дорожки (swimlanes) для отображения ответственности?

6. **Работа с данными**:
   - Какие объекты или данные используются в процессе?
   - Как данные передаются между действиями?

### **6. Диаграмма компонентов**

2. **Определение интерфейсов**:

   - Какие архитектурные паттерны используются в системе?

4. **Анализ производительности и масштабируемости**:

#### **Программные артефакты**



**Программные артефакты** — это все материалы и результаты, которые создаются в процессе разработки программного обеспечения.


##### **Основные виды программных артефактов:**

- **Исходный код** — сами тексты программы, написанные программистами

- **Документация** — описания, инструкции, требования

- **Схемы и диаграммы** — визуальные представления структуры программы

- **Тесты** — материалы для проверки работоспособности программы

- **Исполняемые файлы** — готовые к использованию программы

- **Отчеты об ошибках** — записи о найденных проблемах

Проще говоря, программные артефакты — это все "следы" процесса создания программы, которые помогают понять, как она была создана, как работает и как ее можно улучшить.

5. **Определение условий**:
   - При каких условиях происходит обмен сообщениями?
   - Какие альтернативные пути взаимодействия существуют?

Эта диаграмма помогает заинтересованным сторонам понять масштаб системы и основные функциональные возможности, доступные разным категориям пользователей.

Диаграмма вариантов использования (Use Case Diagram) отображает функциональные требования к системе управления персоналом с точки зрения взаимодействия пользователей с системой.

**Действующие лица (акторы):**
- **Менеджер персонала** - сотрудник отдела кадров, отвечающий за управление данными о персонале
- **Сотрудник** - рядовой работник организации, использующий систему
- **Руководитель** - менеджер, отвечающий за управление подчиненными
- **Администратор** - ИТ-специалист, отвечающий за настройку системы

**Варианты использования:**
1. **Регистрация сотрудника** - процесс добавления нового сотрудника в систему
2. **Просмотр личной информации** - доступ сотрудника к своим данным
3. **Обновление личных данных** - возможность изменения контактной информации
4. **Подача заявления на отпуск** - процесс запроса отпуска
5. **Рассмотрение заявлений** - процесс утверждения/отклонения заявлений подчиненных
6. **Формирование отчетов** - создание аналитических отчетов по персоналу
7. **Управление пользователями** - управление доступом и ролями
8. **Назначение руководителей** - наделение сотрудников руководящими полномочиями

**Отношения:**
- Отношение **include** между "Рассмотрение заявлений" и "Подача заявления на отпуск" означает, что рассмотрение включает в себя получение заявления
- Отношение **extend** между "Назначение руководителей" и "Регистрация сотрудника" показывает, что назначение руководителей может расширить стандартную регистрацию
- Отношение **обобщения** между Руководителем и Сотрудником означает, что руководитель является сотрудником и наследует все его варианты использования



### **2. Диаграмма классов (Class Diagram)**

Диаграмма классов является основой для понимания структуры данных и объектной модели системы, что особенно важно для проектирования базы данных и разработки бизнес-логики приложения.

image.png

Диаграмма классов отображает структуру системы управления персоналом, показывая классы, их атрибуты, методы и отношения между ними.

**Основные классы:**

1. **Person (Человек)** - абстрактный класс, представляющий базовую информацию о человеке
   - Содержит общие атрибуты: идентификатор, имя, дата рождения, контактная информация
   - Предоставляет методы для работы с этими атрибутами

2. **Employee (Сотрудник)** - расширяет класс Person, добавляя специфичные для сотрудника атрибуты
   - Добавляет информацию о найме: ID сотрудника, дата приема, должность, зарплата
   - Содержит связь с отделом (Department) и статус сотрудника
   - Предоставляет методы для управления этими атрибутами

3. **Department (Отдел)** - представляет структурное подразделение компании
   - Содержит идентификатор, название, местоположение
   - Имеет связь с руководителем (Employee) и списком сотрудников
   - Предоставляет методы для управления сотрудниками отдела

4. **Project (Проект)** - представляет проект, над которым работают сотрудники
   - Содержит идентификатор, название, даты начала и окончания, бюджет, статус
   - Имеет список участников проекта (Employee)
   - Предоставляет методы для управления участниками и атрибутами проекта

5. **Перечисления (Enum)**:
   - **EmployeeStatus** - статус сотрудника (активный, в отпуске, уволен, испытательный срок)
   - **ProjectStatus** - статус проекта (планирование, активный, завершен, отменен)

**Отношения между классами:**

1. **Наследование (Inheritance)** - Employee наследует от Person (отношение "является")
   - Сотрудник является человеком и наследует все его атрибуты и методы

2. **Композиция (Composition)** - между Department и Employee (отношение "состоит из")
   - Отдел состоит из сотрудников, причем жизненный цикл сотрудников в контексте отдела зависит от существования отдела

3. **Агрегация (Aggregation)** - между Project и Employee (отношение "содержит")
   - Проект содержит сотрудников как участников, но жизненный цикл сотрудников не зависит от существования проекта

4. **Ассоциация (Association)** - между Department и Employee (связь руководителя)
   - Отдел может иметь одного сотрудника в качестве руководителя

### **4. Диаграмма деятельности (Activity Diagram)**

Эта диаграмма помогает всем участникам процесса понять свои обязанности, последовательность действий и возможные варианты развития событий в зависимости от различных условий.

image.png

Диаграмма деятельности отображает процесс обработки заявления на отпуск, демонстрируя последовательность действий, принятия решений и переходы между участниками процесса.

**Участники процесса (дорожки):**
1. **Сотрудник** - инициатор процесса, запрашивающий отпуск
2. **Система** - автоматизированная система управления персоналом
3. **Руководитель** - менеджер, принимающий решение по заявлению
4. **Отдел кадров** - специалисты, оформляющие документы
5. **Бухгалтерия** - отдел, ответственный за финансовую часть

**Основные этапы процесса:**

1. **Создание заявления:**
   - Сотрудник создает заявление на отпуск
   - Указывает необходимые даты и тип отпуска
   - При необходимости добавляет комментарий
   - Отправляет заявление на рассмотрение

2. **Первичная проверка системой:**
   - Система проверяет корректность указанных дат
   - Если даты некорректны, заявление отклоняется автоматически
   - Система проверяет наличие доступных отпускных дней
   - Если дней недостаточно, заявление также отклоняется
   - В случае успешной проверки, заявление регистрируется и направляется руководителю

3. **Рассмотрение руководителем:**
   - Руководитель получает уведомление о новом заявлении
   - Просматривает детали заявления
   - Проверяет график отпусков отдела для исключения конфликтов
   - Принимает решение об утверждении или отклонении

4. **Обработка решения:**
   - При утверждении:
     - Система обновляет статус заявления и график отпусков
     - Сотрудник получает уведомление об утверждении
   - При отклонении:
     - Руководитель указывает причину отклонения
     - Система обновляет статус заявления
     - Сотрудник получает уведомление с причиной отклонения

5. **Оформление отпуска (для утвержденных заявлений):**
   - Отдел кадров обрабатывает заявление
   - Оформляет необходимые документы (приказ об отпуске)
   - Передает информацию в бухгалтерию

6. **Финансовое оформление:**
   - Бухгалтерия рассчитывает отпускные выплаты
   - Включает их в ближайшую зарплатную выплату

7. **Завершение процесса:**
   - Сотрудник получает отпускные выплаты
   - Уходит в отпуск согласно утвержденным датам

**Особенности диаграммы:**
- **Дорожки (swimlanes)** четко разделяют ответственность участников процесса
- **Условные ветвления** (if-then-else) показывают различные пути выполнения процесса
- **Точки принятия решений** визуализируют основные моменты, влияющие на ход процесса

### **5. Диаграмма состояний (State Diagram)**

Эта диаграмма является важным инструментом как для проектирования системы, так и для документирования бизнес-процессов для конечных пользователей.

image.png

Диаграмма состояний отображает жизненный цикл заявления на отпуск, показывая различные состояния, в которых может находиться заявление, и переходы между этими состояниями.

**Основные состояния:**

1. **Создано** - начальное составное состояние, когда заявление только создано в системе
   - **Черновик** - заявление сохранено как черновик
   - **Отправлено** - заявление подготовлено к отправке

2. **НаРассмотрении** - составное состояние, когда заявление находится в процессе рассмотрения
   - **ОжиданиеРуководителя** - заявление ожидает получения руководителем
   - **РассмотрениеРуководителем** - руководитель рассматривает заявление
   - **ОжиданиеОтделаКадров** - заявление ожидает рассмотрения отделом кадров

3. **Утверждено** - составное состояние, когда заявление утверждено всеми уровнями
   - **ОжиданиеОформления** - ожидание оформления документов
   - **Оформлено** - документы оформлены
   - **ВыполняетсяРасчет** - бухгалтерия проводит расчет отпускных
   - **ГотовоКВыплате** - расчеты завершены, готово к выплате

4. **Отклонено** - заявление отклонено на каком-либо этапе рассмотрения
   - **ОжиданиеОзнакомления** - ожидание ознакомления сотрудника с отказом
   - **Закрыто** - сотрудник ознакомился с причиной отказа

5. **Конечные состояния:**
   - **Отменено** - заявление отменено сотрудником до отправки
   - **Отозвано** - утвержденное заявление отозвано по инициативе компании
   - **Исполнено** - отпуск завершен, все обязательства выполнены
   - **Закрыто** - заявление отклонено и закрыто после ознакомления

**Переходы между состояниями:**

1. **Создание и подготовка:**
   - Заявление создается сотрудником и переходит в состояние "Создано"
   - Внутри "Создано" заявление может быть сохранено как черновик или сразу отправлено
   - Черновик может быть отменен сотрудником или отправлен на рассмотрение

2. **Процесс рассмотрения:**
   - После отправки заявление переходит в состояние "НаРассмотрении"
   - Заявление проходит через этапы ожидания и рассмотрения руководителем
   - По результатам рассмотрения заявление может быть утверждено или отклонено

3. **После утверждения:**
   - Утвержденное заявление переходит в состояние "Утверждено"
   - Происходит оформление документов и расчет отпускных
   - После завершения отпуска заявление переходит в состояние "Исполнено"

4. **Завершение процесса:**
   - Отклоненные заявления требуют ознакомления сотрудника с причиной отказа
   - Завершенные заявления (Отменено, Отозвано, Исполнено, Закрыто) переходят в конечное состояние

**Преимущества использования диаграммы состояний:**
- Дает четкое представление о всех возможных состояниях заявления
- Показывает допустимые переходы между состояниями
- Позволяет выявить логические ошибки в процессе (например, невозможность выхода из какого-либо состояния)
- Помогает разработчикам при проектировании базы данных и бизнес-логики приложения

**Диаграмма классов информационной системы** представляет структуру данных и отношения между основными сущностями системы управления проектами и задачами.

**Основные классы:**

1. **User (Пользователь)** – представляет пользователя системы:

   - Атрибуты: идентификатор, название, описание, даты начала и окончания, статус
   - Операции: управление атрибутами, проверка активности проекта

3. **Task (Задача)** – представляет конкретную задачу в рамках проекта:
   - Атрибуты: идентификатор, заголовок, описание, даты создания и выполнения, приоритет, статус
   - Операции: управление атрибутами, проверка завершенности

4. **Comment (Комментарий)** – содержит комментарии пользователей к задачам:
   - Атрибуты: идентификатор, текст, дата создания
   - Операции: управление содержимым

**Перечисления (Enum):**
- **ProjectStatus** – статусы проекта (планирование, активный, завершенный, архивный)
- **TaskPriority** – приоритеты задач (низкий, средний, высокий, критический)
- **TaskStatus** – статусы задач (к выполнению, в работе, на проверке, выполнена)

**Отношения между классами:**
- **Композиция** (Project *-- Task): проект содержит задачи, которые не могут существовать вне проекта
- **Агрегация** (Task o-- Comment): задача содержит комментарии, которые могут существовать и без задачи
- **Ассоциации** (User -- Project, User -- Task, User -- Comment): показывают связи между пользователями, проектами и задачами
- **Реализация интерфейса** (User --|> Notifiable): пользователь реализует интерфейс для получения уведомлений

### **4. Диаграмма состояний (State Machine Diagram)**

Данная диаграмма состояний наглядно демонстрирует возможные пути, по которым может продвигаться задача от момента создания до завершения, включая обработку исключительных ситуаций, что делает её полезным инструментом для понимания жизненного цикла задачи в системе.

image.png

**Диаграмма состояний жизненного цикла задачи** представляет различные состояния, в которых может находиться задача в процессе её выполнения, и возможные переходы между этими состояниями.

**Основные состояния:**

1. **Создана** – составное состояние, представляющее начальный этап жизни задачи:
   - **Новая** – начальное подсостояние, задача только что создана
   - **Принята** – задача назначена исполнителю и принята им

2. **В работе** – составное состояние, представляющее активную работу над задачей:
   - **Разработка** – основной процесс работы над задачей
   - **Тестирование** – проверка результатов разработки
   - **Исправление** – устранение обнаруженных дефектов

3. **Готова к проверке** – задача завершена и ожидает проверки

4. **Проверка** – составное состояние, представляющее процесс верификации решения:
   - **Код-ревью** – проверка качества кода
   - **Функциональная проверка** – тестирование функциональности

5. **Завершена** – задача полностью выполнена и принята

6. **Отменена** – задача отменена и не будет выполняться

**основные переходы между состояниями:**

1. **Между основными состояниями:**
   - Из состояния "Создана" в "В работе" (startWork) – начало работы над задачей
   - Из "В работе" в "Готова к проверке" (readyForReview) – завершение работы
   - Из "Готова к проверке" в "Проверка" (startReview) – начало проверки
   - Из "Проверка" в "В работе" (returnToWork) – возврат на доработку
   - Из "Проверка" в "Завершена" (approved) – одобрение и завершение
   - В "Отменена" из любого состояния (cancel) – отмена задачи

2. **Внутри составных состояний:**
   - В состоянии "Создана": переход из "Новая" в "Принята" (assignToUser)
   - В состоянии "В работе": цикл "Разработка" -> "Тестирование" -> "Исправление"
   - В состоянии "Проверка": переходы между "Код-ревью" и "Функциональная проверка"

**Особенности диаграммы:**
- Использование составных состояний для группировки логически связанных состояний
- Вложенные машины состояний внутри составных состояний
- Возможность отмены задачи из любого активного состояния
- Циклические переходы внутри состояния "В работе", отражающие итеративный процесс

Диаграмма деятельности представляет процесс анализа кожного заболевания от начала до конца, показывая логический поток действий, решений и параллельные процессы.

**Основные этапы процесса:**

1. **Начало работы:**
   - Пользователь открывает приложение
   - Система проверяет, авторизован ли пользователь
   - Если нет, происходит процесс входа или регистрации

2. **Подготовка к анализу (параллельные действия):**
   - **Получение изображения:**
     - Пользователь выбирает способ получения изображения (галерея или камера)
     - В зависимости от выбора, открывается соответствующий интерфейс
     - Пользователь выбирает существующее фото или делает новый снимок
   
   - **Проверка сетевого подключения:**
     - Система проверяет доступность интернета
     - В зависимости от наличия подключения, подготавливается онлайн или офлайн режим анализа

3. **Обработка и проверка изображения:**
   - Выполняется предварительная обработка изображения
   - Проверяется качество изображения (достаточное разрешение, фокус, освещение)
   - Если изображение не соответствует требованиям, показываются рекомендации по улучшению
   - Если изображение подходит, выполняется его улучшение и выделение области кожи

4. **Классификация заболевания:**
   - В зависимости от режима работы, анализ выполняется онлайн или офлайн
   - В онлайн-режиме изображение отправляется на сервер для анализа более мощной AI-моделью
   - В офлайн-режиме используется локальная модель с меньшей точностью
   - После анализа система получает результат классификации

5. **Обработка результатов:**
   - Результат сохраняется в истории пользователя
   - Одновременно отображаются вероятные заболевания и уровень достоверности результата
   - Пользователю предлагаются возможные дальнейшие действия

6. **Дальнейшие действия (параллельные варианты):**
   - **Просмотр информации о заболевании:**
     - Отображение подробной информации о выявленном заболевании
   
   - **Получение рекомендаций:**
     - Просмотр списка рекомендаций
     - При необходимости, запись к врачу с поиском ближайших специалистов
   
   - **Отмена/Закрытие:**
     - Завершение текущего процесса без дополнительных действий

7. **Завершение:**
   - Возврат на главный экран приложения

**Особенности диаграммы:**
- **Параллельные потоки (fork/join)** показывают действия, которые могут выполняться одновременно
- **Разветвления (if/else)** отображают принятие решений на основе определенных условий
- **Разделение потоков (split)** демонстрирует альтернативные пути, которые может выбрать пользователь
- **Примечания (notes)** предоставляют дополнительную информацию о процессах

Диаграмма деятельности наглядно демонстрирует весь пользовательский путь и логику работы приложения, что помогает понять функциональность и взаимодействие различных компонентов системы.

### **7. Диаграмма развертывания (Deployment Diagram)**

image.png

Диаграмма развертывания показывает физическую архитектуру системы классификации кожных заболеваний, включая аппаратные узлы, программные артефакты и их взаимосвязи.

**Основные узлы системы:**

1. **Облако** - серверная инфраструктура приложения:

   а. **Сервер классификации** - узел, отвечающий за анализ изображений:
      - **API Gateway** - единая точка входа для клиентских запросов
      - **Сервис классификации** - обрабатывает запросы на классификацию изображений
      - **AI модель** - обученная нейронная сеть для распознавания заболеваний кожи

   б. **Сервер базы знаний** - узел, содержащий медицинскую информацию:
      - **Сервис базы знаний** - предоставляет данные о заболеваниях
      - **Сервис рекомендаций** - генерирует персонализированные рекомендации

   в. **Сервер приложения** - узел, обеспечивающий базовую функциональность:
      - **Сервис аутентификации** - управляет входом и безопасностью
      - **Сервис профилей** - управляет пользовательскими данными
      - **Сервис записи к врачу** - обрабатывает запросы на прием к специалистам

   г. **База данных** - централизованное хранилище данных:
      - **Хранилище пользователей** - данные о зарегистрированных пользователях
      - **Хранилище изображений** - загруженные фотографии кожи
      - **Хранилище заболеваний** - информация о кожных заболеваниях
      - **Хранилище врачей** - данные о врачах и их расписаниях

2. **Мобильное устройство** - клиентская часть системы:
   - **Мобильное приложение** - основное приложение для пользователей
   - **Локальная БД** - хранилище данных на устройстве
   - **Легкая AI модель** - упрощенная модель для работы в офлайн-режиме
   - **Кэш изображений** - временное хранилище для обработанных изображений

3. **Внешние сервисы** - сторонние системы для расширения функциональности:
   - **Медицинские API** - доступ к внешним медицинским данным
   - **Платежный сервис** - обработка платежей за консультации
   - **Сервис геолокации** - определение местоположения для поиска врачей

**Основные соединения:**

1. **Внутри облачной инфраструктуры:**
   - Серверы взаимодействуют между собой по протоколу HTTPS
   - Серверы подключаются к базе данных через защищенное SQL-соединение
   - Сервисы используют соответствующие хранилища для чтения и записи данных

2. **Между мобильным устройством и облаком:**
   - Мобильное приложение взаимодействует с серверами через HTTPS
   - Данные передаются в зашифрованном виде

3. **Внутри мобильного устройства:**
   - Приложение использует локальную базу данных для хранения информации
   - В офлайн-режиме приложение использует легкую AI-модель
   - Изображения кэшируются для быстрого доступа и экономии трафика

4. **Взаимодействие с внешними сервисами:**
   - Сервер приложения и мобильное устройство подключаются к внешним сервисам через HTTPS

**Особенности архитектуры:**
- Распределенная облачная архитектура обеспечивает масштабируемость и надежность
- Мобильное приложение имеет возможность работать автономно благодаря локальным компонентам
- Централизованная база данных обеспечивает единую точку доступа к информации
- Интеграция с внешними сервисами расширяет функциональность системы
